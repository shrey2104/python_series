```python
import numpy as np


# This class contains all the functions of our NumPy Analyzer
class DataAnalytics:

    # Constructor
    def __init__(self):
        self.array = None


    # Create a NumPy array
    def create_array(self):

        print("\n--- Create Array ---")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = input("Choose array type: ")

        # Creating 1D array
        if choice == "1":
            values = input("Enter numbers separated by space: ")
            values = list(map(int, values.split()))

            self.array = np.array(values)

        # Creating 2D array
        elif choice == "2":
            rows = int(input("Enter number of rows: "))
            columns = int(input("Enter number of columns: "))

            total = rows * columns

            values = input(f"Enter {total} numbers separated by space: ")
            values = list(map(int, values.split()))

            self.array = np.array(values).reshape(rows, columns)

        # Creating 3D array
        elif choice == "3":
            depth = int(input("Enter depth: "))
            rows = int(input("Enter number of rows: "))
            columns = int(input("Enter number of columns: "))

            total = depth * rows * columns

            values = input(f"Enter {total} numbers separated by space: ")
            values = list(map(int, values.split()))

            self.array = np.array(values).reshape(depth, rows, columns)

        else:
            print("Please enter a valid choice.")
            return

        print("\nArray created successfully:")
        print(self.array)


    # Simple indexing and slicing
    def indexing_slicing(self):

        if self.array is None:
            print("First create an array.")
            return

        print("\nYour Array:")
        print(self.array)

        print("\nFirst element:")
        print(self.array.flat[0])

        print("\nFirst 3 elements:")
        print(self.array.flatten()[:3])


    # Mathematical operations
    def maths_operations(self):

        if self.array is None:
            print("First create an array.")
            return

        print("\n--- Mathematical Operations ---")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")

        choice = input("Enter your choice: ")

        number = float(input("Enter a number: "))

        if choice == "1":
            result = self.array + number

        elif choice == "2":
            result = self.array - number

        elif choice == "3":
            result = self.array * number

        elif choice == "4":

            if number == 0:
                print("Cannot divide by zero.")
                return

            result = self.array / number

        else:
            print("Invalid choice.")
            return

        print("\nResult:")
        print(result)


    # Dot product
    def dot_product(self):

        if self.array is None:
            print("First create an array.")
            return

        if self.array.ndim != 2:
            print("Dot product requires a 2D array.")
            return

        print("\nEnter another matrix.")

        rows = self.array.shape[1]
        columns = int(input("Enter number of columns: "))

        total = rows * columns

        values = input(f"Enter {total} numbers: ")
        values = list(map(int, values.split()))

        second_array = np.array(values).reshape(rows, columns)

        print("\nSecond Array:")
        print(second_array)

        result = np.dot(self.array, second_array)

        print("\nDot Product:")
        print(result)


    # Combine or split arrays
    def combine_split(self):

        if self.array is None:
            print("First create an array.")
            return

        print("\n--- Combine / Split ---")
        print("1. Combine")
        print("2. Split")

        choice = input("Enter your choice: ")

        if choice == "1":

            result = np.concatenate((self.array, self.array))

            print("\nCombined Array:")
            print(result)

        elif choice == "2":

            parts = int(input("How many parts do you want? "))

            result = np.array_split(self.array, parts)

            print("\nSplit Arrays:")

            for part in result:
                print(part)

        else:
            print("Invalid choice.")


    # Search, sort and filter
    def search_sort_filter(self):

        if self.array is None:
            print("First create an array.")
            return

        print("\n--- Search, Sort and Filter ---")
        print("1. Search a value")
        print("2. Sort ascending")
        print("3. Sort descending")
        print("4. Filter values")

        choice = input("Enter your choice: ")

        # Search
        if choice == "1":

            number = int(input("Enter number to search: "))

            position = np.argwhere(self.array == number)

            if len(position) > 0:
                print("Number found at:")
                print(position)
            else:
                print("Number not found.")

        # Ascending sorting
        elif choice == "2":

            result = np.sort(self.array, axis=None)

            print("Ascending order:")
            print(result)

        # Descending sorting
        elif choice == "3":

            result = np.sort(self.array, axis=None)[::-1]

            print("Descending order:")
            print(result)

        # Filtering
        elif choice == "4":

            number = int(input("Show numbers greater than: "))

            result = self.array[self.array > number]

            print("Filtered values:")
            print(result)

        else:
            print("Invalid choice.")


    # Aggregate and statistical functions
    def statistics(self):

        if self.array is None:
            print("First create an array.")
            return

        print("\n--- Array Statistics ---")

        print("Sum:", np.sum(self.array))
        print("Mean:", np.mean(self.array))
        print("Median:", np.median(self.array))
        print("Minimum:", np.min(self.array))
        print("Maximum:", np.max(self.array))
        print("Standard Deviation:", np.std(self.array))
        print("Variance:", np.var(self.array))


    # Correlation between two arrays
    def correlation(self):

        if self.array is None:
            print("First create an array.")
            return

        first_array = self.array.flatten()

        print("\nEnter another array with", len(first_array), "numbers:")

        values = input("Enter numbers: ")
        values = list(map(int, values.split()))

        second_array = np.array(values)

        if len(first_array) != len(second_array):
            print("Both arrays should have the same number of values.")
            return

        result = np.corrcoef(first_array, second_array)

        print("\nCorrelation:")
        print(result)


    # Private method - example of encapsulation
    def __array_details(self):

        print("\n--- Array Details ---")
        print("Shape:", self.array.shape)
        print("Size:", self.array.size)
        print("Dimensions:", self.array.ndim)
        print("Data Type:", self.array.dtype)


    def show_details(self):

        if self.array is None:
            print("First create an array.")
            return

        self.__array_details()


    # Class method
    @classmethod
    def project_info(cls):
        print("\nProject Name: NumPy Analyzer")
        print("This project uses NumPy and OOP concepts.")


    # Static method
    @staticmethod
    def welcome():
        print("\n================================")
        print("Welcome to the NumPy Analyzer!")
        print("================================")


# Main program
analyzer = DataAnalytics()

analyzer.welcome()


while True:

    print("\nChoose an option:")
    print("1. Create a NumPy Array")
    print("2. Mathematical Operations")
    print("3. Dot Product")
    print("4. Combine or Split Arrays")
    print("5. Search, Sort or Filter")
    print("6. Aggregates and Statistics")
    print("7. Indexing and Slicing")
    print("8. Correlation")
    print("9. Show Array Details")
    print("10. Project Information")
    print("11. Exit")

    choice = input("\nEnter your choice: ")


    if choice == "1":
        analyzer.create_array()

    elif choice == "2":
        analyzer.maths_operations()

    elif choice == "3":
        analyzer.dot_product()

    elif choice == "4":
        analyzer.combine_split()

    elif choice == "5":
        analyzer.search_sort_filter()

    elif choice == "6":
        analyzer.statistics()

    elif choice == "7":
        analyzer.indexing_slicing()

    elif choice == "8":
        analyzer.correlation()

    elif choice == "9":
        analyzer.show_details()

    elif choice == "10":
        analyzer.project_info()

    elif choice == "11":
        print("\nThank you for using NumPy Analyzer!")
        break

    else:
        print("Please enter a valid option.")
```
