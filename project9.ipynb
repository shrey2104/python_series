```python
class SalesDataAnalyzer:

    def __init__(self, file_path):
        self.data = pd.read_csv(file_path)
        print("Sales Data Analyzer created successfully!")

    def explore_data(self):
        print(self.data.head())
        print("\nDataset Shape:", self.data.shape)
        print("\nStatistical Summary:")
        print(self.data.describe())

    def clean_data(self):
        self.data.dropna(inplace=True)
        self.data.drop_duplicates(inplace=True)
        print("Data cleaned successfully!")

    def aggregate_functions(self):
        print("Total Sales:", self.data["SALES"].sum())
        print("Average Sales:", self.data["SALES"].mean())
        print("Total Profit:", self.data["PROFIT"].sum())

    def statistical_analysis(self):
        print("Mean:", self.data["SALES"].mean())
        print("Standard Deviation:", self.data["SALES"].std())
        print("Variance:", self.data["SALES"].var())

    def create_pivot_table(self):
        pivot = pd.pivot_table(
            self.data,
            values="SALES",
            index="PRODUCT",
            columns="REGION",
            aggfunc="sum",
            fill_value=0
        )

        print(pivot)

    def __del__(self):
        print("Sales Data Analyzer closed.")
```

```python
analyzer = SalesDataAnalyzer("sales_data.csv")
```

```python
analyzer.explore_data()
```

```python
analyzer.clean_data()
```

```python
analyzer.aggregate_functions()
```

```python
analyzer.statistical_analysis()
```

```python
analyzer.create_pivot_table()
```
