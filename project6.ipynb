```python
import os
from datetime import datetime


# Class to manage our personal journal
class JournalManager:

    def __init__(self):
        self.filename = "journal.txt"

    # 1. Add a new journal entry
    def add_entry(self):

        entry = input("\nEnter your journal entry: ")

        if entry.strip() == "":
            print("Journal entry cannot be empty.")
            return

        # Get current date and time
        date = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        try:
            # Append mode adds new data without deleting old entries
            with open(self.filename, "a") as file:
                file.write(f"[{date}]\n")
                file.write(entry + "\n\n")

            print("Entry added successfully!")

        except PermissionError:
            print("Error: Permission denied. Cannot write to file.")

        except OSError as e:
            print("File error:", e)


    # 2. View all journal entries
    def view_entries(self):

        try:
            with open(self.filename, "r") as file:
                data = file.read()

            if data.strip() == "":
                print("Your journal is empty.")
            else:
                print("\nYour Journal Entries:")
                print("-" * 40)
                print(data)

        except FileNotFoundError:
            print("Error: The journal file does not exist.")
            print("Please add a new entry first.")

        except PermissionError:
            print("Error: Permission denied. Cannot read the file.")

        except OSError as e:
            print("File error:", e)


    # 3. Search journal entry
    def search_entry(self):

        keyword = input("\nEnter a keyword or date to search: ")

        if keyword.strip() == "":
            print("Please enter a valid keyword.")
            return

        try:
            with open(self.filename, "r") as file:
                data = file.read()

            # Separate individual journal entries
            entries = data.strip().split("\n\n")

            found = False

            for entry in entries:

                if keyword.lower() in entry.lower():

                    if found == False:
                        print("\nMatching Entries:")
                        print("-" * 40)

                    print(entry)
                    print()
                    found = True

            if found == False:
                print("No entries were found for the keyword:", keyword)

        except FileNotFoundError:
            print("Journal file not found. Please add an entry first.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("File error:", e)


    # 4. Delete all journal entries
    def delete_entries(self):

        if not os.path.exists(self.filename):
            print("No journal entries to delete.")
            return

        confirm = input(
            "\nAre you sure you want to delete all entries? (yes/no): "
        )

        if confirm.lower() == "yes":

            try:
                os.remove(self.filename)
                print("All journal entries have been deleted.")

            except PermissionError:
                print("Error: Permission denied. Cannot delete the file.")

            except FileNotFoundError:
                print("The journal file does not exist.")

            except OSError as e:
                print("File error:", e)

        else:
            print("Delete operation cancelled.")


# Main program
def main():

    # Creating an object of JournalManager class
    journal = JournalManager()

    while True:

        print("\n================================")
        print("Welcome to Personal Journal Manager!")
        print("================================")

        print("1. Add a New Entry")
        print("2. View All Entries")
        print("3. Search for an Entry")
        print("4. Delete All Entries")
        print("5. Exit")

        choice = input("\nPlease select an option: ")

        if choice == "1":
            journal.add_entry()

        elif choice == "2":
            journal.view_entries()

        elif choice == "3":
            journal.search_entry()

        elif choice == "4":
            journal.delete_entries()

        elif choice == "5":
            print("\nThank you for using Personal Journal Manager. Goodbye!")
            break

        else:
            print("Invalid option. Please select a valid option from the menu.")


# Run the program
if __name__ == "__main__":
    main()
```
