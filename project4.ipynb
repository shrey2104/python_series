```python

# ==========================================
# FUNCTIONAL TREAT PROJECT
# Data Analyzer and Transformer
# ==========================================

operations = 0

# -----------------------------
# Sample Data
# -----------------------------
sample_1d = [45, 12, 78, 23, 89, 5, 67]
sample_2d = [
    [23, 56, 78],
    [12, 98, 45],
    [67, 34, 89]
]


# -----------------------------
# Global Counter
# -----------------------------
def count_operation():
    """Counts total operations performed."""
    global operations
    operations += 1


# -----------------------------
# Input Functions
# -----------------------------
def get_1d_list():
    """Accept 1D list from user."""
    n = int(input("Enter number of elements: "))
    arr = []

    for i in range(n):
        arr.append(int(input(f"Element {i+1}: ")))

    return arr


def get_2d_list():
    """Accept 2D list from user."""
    rows = int(input("Rows: "))
    cols = int(input("Columns: "))

    matrix = []

    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input(f"Element [{i}][{j}]: ")))
        matrix.append(row)

    return matrix


# -----------------------------
# Analysis Functions
# -----------------------------
def analyze(arr):
    """Returns sum, average, maximum and minimum."""
    count_operation()

    total = sum(arr)
    average = total / len(arr)
    maximum = max(arr)
    minimum = min(arr)

    return total, average, maximum, minimum


# -----------------------------
# Sorting
# -----------------------------
def sort_list(arr):
    """Returns ascending and descending list."""
    count_operation()

    asc = sorted(arr)
    desc = sorted(arr, reverse=True)

    return asc, desc


# -----------------------------
# Lambda Function
# -----------------------------
def square_elements(arr):
    """Squares each element using lambda."""
    count_operation()

    return list(map(lambda x: x ** 2, arr))


# -----------------------------
# *args Example
# -----------------------------
def total_sum(*numbers):
    """Find sum using *args."""
    count_operation()

    return sum(numbers)


# -----------------------------
# **kwargs Example
# -----------------------------
def student_info(**details):
    """Display keyword arguments."""
    count_operation()

    print("\nStudent Details")
    for key, value in details.items():
        print(key, ":", value)


# -----------------------------
# Recursion
# -----------------------------
def recursive_sum(arr, n):
    """Recursive sum of list."""

    if n == 0:
        return 0

    return arr[n - 1] + recursive_sum(arr, n - 1)


# -----------------------------
# 2D Functions
# -----------------------------
def transpose(matrix):
    """Transpose matrix."""
    count_operation()

    return [list(row) for row in zip(*matrix)]


def flatten(matrix):
    """Convert 2D list into 1D list."""
    count_operation()

    return [item for row in matrix for item in row]


# -----------------------------
# Documentation
# -----------------------------
def show_docs():
    print("\nFunction Documentation")
    print(analyze.__doc__)
    print(sort_list.__doc__)
    print(square_elements.__doc__)
    print(total_sum.__doc__)
    print(student_info.__doc__)
    print(recursive_sum.__doc__)
    print(transpose.__doc__)
    print(flatten.__doc__)


# -----------------------------
# Main Program
# -----------------------------
while True:

    print("\n" + "=" * 50)
    print("FUNCTIONAL TREAT PROJECT")
    print("=" * 50)

    print("1. Use Sample 1D Data")
    print("2. Enter Your Own 1D Data")
    print("3. Use Sample 2D Data")
    print("4. Enter Your Own 2D Data")
    print("5. Show Function Documentation")
    print("6. Show Total Operations")
    print("7. Exit")

    choice = input("Enter choice: ")

    if choice == "1":

        arr = sample_1d

        print("\nList:", arr)

        total, avg, maximum, minimum = analyze(arr)

        print("Sum =", total)
        print("Average =", avg)
        print("Maximum =", maximum)
        print("Minimum =", minimum)

        asc, desc = sort_list(arr)

        print("Ascending =", asc)
        print("Descending =", desc)

        print("Squares =", square_elements(arr))

        print("Recursive Sum =", recursive_sum(arr, len(arr)))

        print("*args Sum =", total_sum(*arr))

        student_info(Name="Aanal", Course="Data Analysis", Semester=5)

    elif choice == "2":

        arr = get_1d_list()

        total, avg, maximum, minimum = analyze(arr)

        print("\nSum =", total)
        print("Average =", avg)
        print("Maximum =", maximum)
        print("Minimum =", minimum)

        asc, desc = sort_list(arr)

        print("Ascending =", asc)
        print("Descending =", desc)

        print("Squares =", square_elements(arr))

        print("Recursive Sum =", recursive_sum(arr, len(arr)))

    elif choice == "3":

        matrix = sample_2d

        print("\nMatrix")

        for row in matrix:
            print(row)

        print("\nTranspose")

        for row in transpose(matrix):
            print(row)

        print("\nFlattened List")

        print(flatten(matrix))

    elif choice == "4":

        matrix = get_2d_list()

        print("\nOriginal Matrix")

        for row in matrix:
            print(row)

        print("\nTranspose")

        for row in transpose(matrix):
            print(row)

        print("\nFlattened List")

        print(flatten(matrix))

    elif choice == "5":
        show_docs()

    elif choice == "6":
        print("\nTotal Operations Performed =", operations)

    elif choice == "7":
        print("\nThank You!")
        break

    else:
        print("\nInvalid Choice!")
```
